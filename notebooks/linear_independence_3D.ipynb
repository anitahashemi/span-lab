```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

project_root = Path.cwd()
if project_root.name == "notebooks":
    project_root = project_root.parent

figures_dir = project_root / "figures"
figures_dir.mkdir(parents=True, exist_ok=True)
```

### Defining Vectors 

```python
u = np.array([1, 0, 1])
v = np.array([0, 1, 1])
w = np.array([1, 1, 0])
# For dependent vector you should try: w = u + v.

A = np.column_stack([u, v, w])

print(f"Determinant: {np.linalg.det(A):.2f}")
print(f"Rank: {np.linalg.matrix_rank(A)}")
```

### Adding Visualization 

```python
coefficients = np.linspace(-1, 1, 7)

points = np.array([
    a * u + b * v + c * w
    for a in coefficients
    for b in coefficients
    for c in coefficients
])

fig = plt.figure(figsize=(8, 7))
ax = fig.add_subplot(111, projection="3d")

ax.scatter(
    points[:, 0],
    points[:, 1],
    points[:, 2],
    color="teal",
    alpha=0.25,
    s=12,
    label="Sampled linear combinations"
)

for vector, color, label in [
    (u, "royalblue", "u"),
    (v, "darkorange", "v"),
    (w, "purple", "w"),
]:
    ax.quiver(
        0, 0, 0,
        *vector,
        color=color,
        linewidth=2.5,
        arrow_length_ratio=0.15,
        label=label
    )
    ax.text(*(1.15 * vector), label, color=color, fontsize=13)

limit = np.max(np.abs(points)) + 0.5
ax.set_xlim(-limit, limit)
ax.set_ylim(-limit, limit)
ax.set_zlim(-limit, limit)
ax.set_box_aspect((1, 1, 1))

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
ax.view_init(elev=22, azim=35)
ax.legend()

rank = np.linalg.matrix_rank(A)
ax.set_title(f"Linear combinations in 3D | rank = {rank}")

filename = (
    "independent_vectors_3d.png"
    if rank == 3
    else "dependent_vectors_3d.png"
)
fig.savefig(figures_dir / filename, dpi=150, bbox_inches="tight")
plt.show()
```
